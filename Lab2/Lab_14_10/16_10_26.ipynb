```python
from langchain_groq import ChatGroq
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
input_temp = ChatPromptTemplate.from_messages([
    ("system","You are a experience poet . Write meaningful poem. Do not response anything outside of poem"),
    ("user","Write Poem About {poem_topic}")
])
```

```python
input = input_temp.invoke(("poem_topic:rain fall"))
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)
```

```python
response = llm.invoke(input)
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser = StrOutputParser()
```

```python
parser.invoke(response)
```

```python
input = input_temp.invoke({"poem_topic":"rain fall"})
responses = llm.invoke(input)
final_responses = parser.invoke(responses)
```

```python
print(final_responses)
```

```python
chain = input_temp | llm |parser
```

```python
chain.invoke({"poem_topic":"rain fall"})
```

```python
json_parser = StrOutputParser()
```

```python
from langchain_core.output_parsers import JsonOutputParser

json_parser = JsonOutputParser()

format_instructions = json_parser.get_format_instructions()
print(format_instructions)
```

```python
from langchain_core.prompts import ChatPromptTemplate

summarize_temp = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are an expert in summarizing poems. "
        "Summarize the poem in 5 points.\n{JSON_FORMATTING}"
    ),
    ("user", "Summarise this poem: {poem}")
]).partial(
    JSON_FORMATTING=json_parser.get_format_instructions()
)
```

```python
summarize_temp 
```

```python
summarize_temp.invoke({"poem":"final_responses"})
```

```python
step_2_input = summarize_temp.invoke({
    "poem": "final_response"
})

print(step_2_input)
```

```python
llm.invoke(step_2_input)
```

```python
step_2_resp = llm.invoke(step_2_input)
```

```python
json_parser.invoke(step_2_resp)
```
